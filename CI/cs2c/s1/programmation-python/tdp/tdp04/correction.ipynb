#### Exercice 1: **Supprimer les doublons d'une liste**

- Générer une liste L de 40 entiers entre 8 et 24

```python
import random
# pour generer un nombre entier aleatoirement entre Vmin et Vmax
Vmin = 8 # Ex
Vmax = 24 # Ex
L = [random.randint(Vmin,Vmax) for _ in range(40)]
print(L)
```

- Ecrire une fonction qui affiche le nombre le plus fréquent 

```python
def nombre_plus_frequent(liste):
    if not liste:
        print("La liste est vide.")
        return

    frequences = {}
    for item in liste:
        frequences[item] = frequences.get(item, 0) + 1
    
    num_max = max(frequences, key=frequences.get)
    freq_max = frequences[num_max]
    
    print(f"Le nombre le plus frequent {num_max}, son occurence est {freq_max}.")

nombre_plus_frequent(L)
```

- Ecrire une fonction qui supprime tous les doublons de la liste L. Vous pouvez utiliser un espace auxiliaire.

```python
def supprimer_doublons(liste):
    liste_sans_doublons = []
    for item in liste:
        if item not in liste_sans_doublons:
            liste_sans_doublons.append(item)
    return liste_sans_doublons

L_sans_doublons_1 = supprimer_doublons(L)
print(L_sans_doublons_1)
```

- Refaire la question en utilisant la notion de set. 

```python
def supprimer_doublons_set(liste):
    return list(set(liste))
    
L_sans_doublons_2 = supprimer_doublons_set(L)
print(sorted(L_sans_doublons_2))
```

- Remarques & Conclusions ?

On remarque que :
- La première méthode garde l'ordre des éléments, la méthode avec le set ne le garde pas.
- La première méthode est en O(n²) : `item not in liste_sans_doublons` parcourt la liste à chaque tour.
- La méthode avec le set est en O(n) en moyenne, donc beaucoup plus rapide sur de grandes listes.

-	Écrire une fonction qui prend deux sets et retourne leur intersection (éléments communs).


```python
set_A = {random.randint(1, 20) for _ in range(15)}
set_B = {random.randint(1, 20) for _ in range(15)}
print(set_A)
print(set_B)

def intersection(s1, s2):
    resultat = set()
    for element in s1:
        if element in s2:
            resultat.add(element)
    return resultat

print(intersection(set_A, set_B))


print(f"Intersection : {set_A & set_B}")
```

- Écrire la fonction FrqLet qui prend en paramètre une chaine et retourne un dictionnaire des fréquences de chaque lettre de la chaine

```python
def FrqLet(ch):
    # ecrire votre solution ici
    frequences = {}
    for lettre in ch:
        frequences[lettre] = frequences.get(lettre, 0) + 1
    return frequences
```

```python
D1 = FrqLet(ch1)
D2 = FrqLet(ch2)
print(D1)
print(D2)

def fusionner_somme(d1, d2):

    d3 = d1.copy()
    for cle, valeur in d2.items():
        d3[cle] = d3.get(cle, 0) + valeur
    return d3

D3 = fusionner_somme(D1, D2)
print(D3)
```

- Écrire une fonction qui inverse les clés et les valeurs d'un dictionnaire (clé ↔ valeur). <br>
Si plusieurs clés ont la même valeur, stocker les clés dans une liste.<br>
Exemple : {'a': 1, 'b': 2, 'c': 1} → {1: ['a', 'c'], 2: ['b']}

```python
def invDic(di):
    d_inverse = {}
    for cle, valeur in di.items():
        d_inverse.setdefault(valeur, []).append(cle)
    return d_inverse

d_exemple = {'a': 1, 'b': 2, 'c': 1, 'd': 3, 'e': 2}
d_inverse_exemple = invDic(d_exemple)
print(f"\norginale: {d_exemple}")
print(f"Inverse: {d_inverse_exemple}")
```

- Générer aléatoirement une liste LP de 300 points dont les coordonnés (x,y) sont des entiers (relatifs) entre -15 et 15 

```python
import math

LP = [(random.randint(-15, 15), random.randint(-15, 15)) for _ in range(300)]

print(LP)
print(f"Nombre de points : {len(LP)}")
```

À partir de la liste de LP, créer un dictionnaire DP dont la clé est le tuple (x, y) et la valeur est la distance euclidienne à l’origine. <br>
**N.B.** le dictionnaire DP ne doit pas contenir de **doublons**

```python
DP = {
    point: math.sqrt(point[0]**2 + point[1]**2) 
    for point in set(LP)
}

print(f"Nombre de points : {len(DP)}")
```

- écrire une fonction **PtInf** qui retourne tous les points de DP qui sont à une distance à P inférieure à d 
    - Le dictionnaire D, la distannce d et le point P sont des paramètres de la fonction

```python
def PtInf(D , d, P = (0,0)):
    points_proches = []
    px, py = P
    for point_courant in D.keys():
        cx, cy = point_courant
        distance = math.sqrt((cx - px)**2 + (cy - py)**2)
        if distance < d:
            points_proches.append(point_courant)
            
    return points_proches
    
points_proches_origine = PtInf(DP, 5)
print(f"\nPoints proche de l'orgine : {points_proches_origine}")

point_ref = (10, 10)
points_proches_de_P = PtInf(DP, 4, P=point_ref)
print(f"Points proche de P {point_ref}: {points_proches_de_P}")
```
