### 1. Expression régulière

Les expressions régulières (ou "regex") sont des chaînes de caractères utilisées pour rechercher, extraire, ou valider des **motifs** (patterns) dans du texte (ex. : email, numéro de téléphone, mot-clé...).

Python fournit le module **re** pour les manipuler (import re)

```python
import re
```

- Syntaxe des motifs (patterns)

| Motif   | Signification                                   | Exemple(s) d’utilisation                                               |
| ------- | ----------------------------------------------- | ---------------------------------------------------------------------- |
| `.`     | N’importe quel caractère sauf retour à la ligne | `a.c` : correspond à `abc`, `axc` mais pas `ac`                        |
| `^`     | Début de ligne                                  | `^Hello` : correspond à `Hello world` si la ligne commence par `Hello` |
| `$`     | Fin de ligne                                    | `end$` : correspond à `the end`                                        |
| `[]`    | Classe de caractères (ensemble autorisé)        | `[0-9]` : un chiffre, `[aeiou]` : une voyelle                          |
| `\|`    | OU logique (alternatives possibles)             | `chat\|chien` : correspond à `chat` ou `chien`                         |

- Motifs de base

| Pattern | Signification                                | Exemple                             |
| ------- | -------------------------------------------- | ----------------------------------- |
| `.`     | N'importe quel caractère (sauf retour ligne) | `a.c` correspond à `"abc"`, `"a7c"` |
| `\d`    | Un chiffre (0-9)                             | `\d\d` correspond à `"42"`          |
| `\D`    | Tout sauf un chiffre                         | `\D+` correspond à `"abc"`          |
| `\w`    | Un caractère alphanumérique ou `_`           | `\w+` → `"Nom123"`                  |
| `\W`    | Tout sauf un caractère alphanumérique        | `\W+` → `"@#!"`                     |
| `\s`    | Un espace, tabulation ou retour ligne        | `\s+` → `"  "`                      |
| `\S`    | Tout sauf un espace                          | `\S+` → `"Texte"`                   |


- Quantificateurs:

| Pattern | Signification     | Exemple                       |
| ------- | ----------------- | ----------------------------- |
| `*`     | 0 ou plusieurs    | `ab*` : `"a"`, `"abbb"`       |
| `+`     | 1 ou plusieurs    | `ab+` : `"ab"`, `"abbbb"`     |
| `?`     | 0 ou 1 fois       | `ab?` : `"a"`, `"ab"`         |
| `{n}`   | exactement n fois | `\d{4}` : `"2025"`            |
| `{n,}`  | au moins n fois   | `\d{2,}` : `"1234"`           |
| `{n,m}` | entre n et m fois | `\d{2,4}` : `"123"`, `"2025"` |


- Délimiteurs et classes de caractères

| Pattern    | Signification                 | Exemple                         |
| ---------- | ----------------------------- | ------------------------------- |
| `[...]`    | Une des valeurs listées       | `[abc]` → `"a"`, `"b"` ou `"c"` |
| `[^...]`   | Tout sauf les valeurs listées | `[^abc]` → `"d"`, `"z"`         |
| `[0-9]`    | Chiffre de 0 à 9              | `[0-9]+` → `"2025"`             |
| `[a-zA-Z]` | Lettre majuscule ou minuscule | `[a-zA-Z]+` → `"Python"`        |


- Début et fin de chaîne

| Pattern | Signification   | Exemple                                             |
| ------- | --------------- | --------------------------------------------------- |
| `^`     | Début de chaîne | `^Bonjour` → correspond à `"Bonjour Alice"`         |
| `$`     | Fin de chaîne   | `2025$` → correspond à `"année 2025"`               |
| `\b`    | Limite de mot   | `\bmot\b` → correspond à `"mot"` mais pas `"motif"` |


- Exemples de motifs utiles

| Objectif                     | Regex                      |
| ---------------------------- | -------------------------- |
| Nombre entier                | `^\d+$`                    |
| Date (format jj/mm/aaaa)     | `\d{2}/\d{2}/\d{4}`        |


- Fonctions principales de **re**

| Fonction        | Rôle                                                             |
| --------------- | ---------------------------------------------------------------- |
| `re.search()`   | Recherche un motif **une seule fois** dans une chaîne            |
| `re.match()`    | Teste si le **début** de la chaîne correspond au motif           |
| `re.findall()`  | Retourne **toutes** les correspondances                          |
| `re.finditer()` | Idem que `findall` mais retourne un **itérateur d'objets Match** |
| `re.sub()`      | Remplace des parties de texte selon un motif                     |
| `re.split()`    | Coupe une chaîne selon un motif                                  |


### Exemples

- search()

```python
# re.search() : Recherche un motif une seule fois dans une chaîne
import re

# Exemple 1 : présence d'un mot
texte = "Bienvenue au cours Python"
corresp = re.search(r"Python", texte)
print(corresp.group())  

# Exemple 2 : chiffre dans une phrase
corresp = re.search(r"\d+", "Il a 13 ans 15 ans")
print(corresp.group())  # 13

# Exemple 3 : Non chiffre dans une phrase
corresp = re.search(r"\D+", "Il a 13 ans")
print(corresp.group())  
```

- match()

```python
import re
# re.match() : Vérifie si le début de la chaîne correspond au motif.
# Exemple 1 : bon début
resultat = re.match(r"Bonjour", "Bonjour tout le monde")
print(resultat, "de type ", type(resultat)) 

print(resultat.group())  # .group() sert à récupérer le texte exact trouvé par la correspondance (match) : Bonjour

# Exemple 2 : ne matche pas car ce n’est pas au début
print(re.match(r"tout", "Bonjour tout le monde"))  # None
```

- findall()

```python
# re.findall() : Retourne toutes les correspondances sous forme de liste.
# Exemple 1 : trouver tous les chiffres
nums = re.findall(r"\d+", "Année 2020, puis 2024, et maintenant 2025")
print(nums)  # ['2020', '2024', '2025']

# Exemple 2 : mots commençant par une majuscule
mots = re.findall(r"\b[A-Z]\w+", "Alice est avec C Bob et Oscar")
print(mots)  # ['Alice', 'Bob', 'Oscar']
```

- re.finditer()

```python
# re.finditer() : Comme findall, mais retourne des objets Match, donc plus d’info
# Exemple 1 : trouver les positions des chiffres
for m in re.finditer(r"\d+", "123 et 456"):
    print(f"Nombre : {m.group()}, position : {m.start()}")  
# Nombre : 123, position : 0
# Nombre : 456, position : 7

# Exemple 2 : toutes les majuscules avec leur position
texte = "Python Et C"
for match in re.finditer(r"[A-Z]", texte):
    print(match.group(), match.start())
# P 0
# E 7
# C 10
```

- split()

```python
# re.split()
# Exemple 1 : couper par espace ou virgule
print(re.split(r"[ ,]+", "Python, PHP Java,, C++")) 
# ['Python', 'PHP', 'Java', 'C++']

# Exemple 2 : couper à chaque chiffre
print(re.split(r"\d+", "abc123def456ghi")) # ['abc', 'def', 'ghi']
```

- sub()

```python
# re.sub(): Remplace les correspondances par une nouvelle valeur.
# Exemple 1 : anonymiser un email
texte = "Mon email est alice@gmail.com"
print(re.sub(r"\S+@\S+", "[EMAIL]", texte)) # Mon email est [EMAIL]

# Exemple 2 : remplacer tous les chiffres par '*'
print(re.sub(r"\d", "*", "Code 12345"))   # Code *****
```

- fullmatch()<br>
La fonction re.fullmatch() vérifie si toute la chaîne correspond exactement au motif.

```python
import re

motif = r"\d+"  # Un ou plusieurs chiffres

print(re.fullmatch(motif, "12345"))    #  OK → Match

print(re.fullmatch(motif, "123abc"))   #  NON → None
```

#### - Méthodes importantes
- **ravel()** : pour applatir un tableau (qu'il ne fasse plus qu'une dimension)
- **squeeze()** : quand une dimension est égale a 1, cette dimension disparait

```python
import numpy as np
A = np.zeros((2, 3)) # création d'un tableau de shape (2, 3)
print("-------------- A")
print(A)
```

```python
print("-------------- A.reshape()")
A = A.reshape((3, 2)) # redimensionne le tableau A (3 lignes, 2 colonnes)
print(A)
```

```python
print("-------------- A.ravel()")
X = A.ravel() # Aplatit le tableau A (une seule dimension)
print(X) 
```

```python
Y = np.array([[[1, 1, 1], [2, 4, 8]]]) 
print ("- Y : \n", Y) 

d1 = 0
print (f"\n- Y[{d1}] :\n {Y[d1]}")  

d2 = 1
print (f"\n- Y[{d1},{d2}] :\n {Y[d1,d2]}")  

d3 = 2
print (f"\n- Y[{d1},{d2},{d3}] :\n {Y[d1,d2,d3]}")  

print("\nShape de Y : ", Y.shape)   
```

```python
import time
import random
nbf = 1_000_000
a = np.array([random.randint(0,10) for i in range(nbf)])
b = a.copy()
```

```python
deb = time.time()
a.sort()
fin = time.time()

print("temps sort() :",fin-deb)
```

```python
deba = time.time()
pos = b.argsort()
vtr = b[pos]
fina = time.time()
print("temps argsort() :",fina-deba)
```

```python
import numpy as np
```

```python
A = np.array([np.arange(5*i, 5*i+3) for i in range(2)])
print(A)
```

```python
B = 3
print("A :\n",A)
print("A+B :")
print(A+B) # résultat 1
```

```python
C = np.array([[3],[4]])

print("A:\n", A)
print("C:\n", C)
print(C.shape)
```

```python
print("A+C:")
print(A+C)
```
