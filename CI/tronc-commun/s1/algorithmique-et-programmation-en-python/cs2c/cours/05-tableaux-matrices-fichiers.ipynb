## Modifier les éléments d'une matrice

## Produits matriciels de A et B:
    - A@B 
    - np.dot(A,B)
    - A*B : Element By Element
    

```python
import numpy as np

A = np.array([[10, 20, 70], [30, 40, 90], [100, 200, 300]])
B = np.array([[1, 2, 3], [6, 5, 4], [7, 9, 8]])
# B = np.array([[10, 20], [30, 40], [100, 200]])

print(f"\nA {np.shape(A)}: \n", A)
print(f"\nB {np.shape(B)}: \n", B)

AT = A@B
print(f"\nA@B : {np.shape(AT)}: \n", AT)

DOT = np.dot(A,B)
print(f"\ndot : {np.shape(DOT)}: \n", DOT)
```

```python
Res = (DOT == AT)
print("Produit .dot() == @ :", Res.all())
```

##  Transposée d'une Matrice

```python
import numpy as np

Mat = np.array([[10, 20, 70], [30, 40, 90]])

TrMat = Mat.transpose()
print(f"Matrice Transpose  : {np.shape(TrMat)}\n",TrMat)

TrsMat2 = Mat.T
print(f"Matrice 2 Transpose : {np.shape(TrMat)}\n",TrsMat2)
```

```python
ResTr = (TrMat == TrsMat2)
print("transposee .transpose() == .T :", ResTr.all())
```

## Déterminant - numpy.linalg.det()

```python
from numpy.linalg import det
mt = np.array([[2, 2],
                 [3, 4]])

print(det(mt))
print(round(det(mt)))
# ------------ !!! ATTENTION : ne pas utiliser int() !!!
print(int(det(mt))) 
```

```python
def deter(a):
    return a[0][0]*a[1][1] - a[1][0]*a[0][1] 

print(deter(mt))
```

# <font color=blue> Fichiers : </font>

- __Python dispose de plusieurs fonctions pour créer, lire et mettre à jour des fichiers__.

- La fonction clé pour travailler avec des fichiers en Python est la fonction open().

- La fonction open() prend deux paramètres : le nom du fichier et le mode.

- Il existe quatre méthodes différentes (modes) pour ouvrir un fichier :

    - "r" - Read - Valeur par défaut. Ouvre un fichier en lecture, erreur si le fichier n'existe pas.

    - "a" - Append - Ouvre un fichier pour l'ajouter, crée le fichier s'il n'existe pas.

    - "w" - Write - Ouvre un fichier pour l'écriture, crée le fichier s'il n'existe pas.

    - "x" - Create - Crée le fichier spécifié, renvoie une erreur si le fichier existe.

- En outre, vous pouvez spécifier si le fichier doit être traité en mode binaire ou en mode texte

    - "t" - Texte - Valeur par défaut. Mode texte

    - "b" - Binaire - Mode binaire (par exemple, images)


## <font color=red>ATTENTION </font> : à chaque <font color=red>  open() </font> il faut associer automatiquement un <font color=red>close() </font>

## Ouverture en lecture "r"

### Lire tout le fichier d'un seul coup

```python
f = open("etudiants.txt")
# f = open("demofile.txt", "rt")

print("Debut")
print(f.read()) # affiche tout le fichier
print("Fin")
f.close()
```

### Lire le fichier Ligne par Ligne

```python
f = open("etudiants.txt", "rt")

for x in f:
    print("New Line : ", end = "")
    print(x)
f.close()
```

```python
f = open("etudiants.txt")

x = f.readline()
while(x):
    print(x)
    x = f.readline()
f.close()
```

### Ouverture en ajout "a"


```python
f = open("etudiants.txt" , "a")
f.write("\nHind:23:F:5:Oujda")
f.close()

f = open("etudiants.txt")
print(f.read()) # affiche tout le fichier
f.close()
```

## Ouverture en écriture "w" (écrase l'existant)

```python
f = open("etudiants.txt", "w")
f.write("Fatal Error ! I have deleted the content !")
f.close()


f = open("etudiants.txt")
print(f.read()) # affiche tout le fichier
f.close()
```
