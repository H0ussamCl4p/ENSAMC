## Opérateurs

- Les opérateurs sont des symboles spéciaux utilisés pour effectuer des opérations sur des variables et des valeurs.

- Python supporte plusieurs catégories : (avec exemples)
    - Arithmétiques : a + 3, ...
    - Relationnels : x == y
    - Logiques : p and q
    - D’affectation : a = 7
    - Bit à bit : a & b
    - Spéciaux : in

#### Operateurs Unaires : s'appliquent à un seul opérande :: exemples

```python
# not :
a = 6
if (not a):
    print("Tr a")
else:
    print("False a")
```

```python
# moins :
b = 6
print(-b)
```

```python
# Complément bit à bit :
x = 5          # 5 en binaire: 00000101
print(~x)      # -6 (complément: 11111010)
```

... à voir dans la suite du cours : in, is ... etc

#### Operateurs Binaires : s'appliquent à deux opérandes :: exemples

```python
a = 10
b = 3

print(a + b)   # 13
print(a > b)   # True
```

## Expression

- un ensemble de valeur(s), 
- reliées éventuellement par des opérateurs,  
- équivalent à une seule valeur:
- exemples :
    * a + B
    * "You get what you work for, not what you wish for"
    * som  !=  A + b 
    * E

- Valeur **numérique** d'une expression (numerique) :  valeur arithmétique
<br>Exemples: 
    - si une variable x vaut 7, alors sa valeur numérique est 7
    - La valeur numérique de l'expression de l'affectation : s = 13 est égale à 13

- Valeur **logique** d'une expression: valeur booléenne
- Deux possibilités:
    * Vrai ou 1 : la valeur logique d'une expression est "Vrai" si sa valeur numérique est différente de 0, <br>(y compris les valeurs négatives)
    * Faux ou 0: la valeur logique d'une expression est "Faux" si sa valeur numérique est égale à 0

```python
if (14-14) :
    print ("T")
else:
    print("Fal")
```

```python
ch = "Alphabet"
cha = ""

if(ch):
    print("chaine NON vide : valeur True")
else:
    print("chaine NON vide : valeur False")

if(cha):
    print("chaine vide : valeur True")
else:
    print("chaine vide : valeur False")
```

### Attention aux floats : Erreur dûe au codage
 

```python
print((3 - 2.8) == 0.2)

print(3 - 2.8)
```

#### Alternative :

```python
delta = 0.0001
var = 3.0 - 2.8
print(0.2 - delta < var < 0.2 + delta)
print (abs(var - 0.2) < delta)
```

```python
print(0.3)
print(f"{0.3:.5f}")
print(f"{0.3:.30f}")
print(f"{3 - 2.7:.30f}")

# On observe que lorsqu'on tape 0.3, Python affiche une valeur arrondie. 
# En se limitant à une quantité finie de bits, on ne peut obtenir qu'une approximation.
# le nombre réel 0.3 ne peut être qu'approché lorsqu'on le code en nombre flottant. 
```
