# <font color=blue>Tuples :</font> <font size=3>(Fr : n-uplets)</font>
<pre>
- Les tuples sont écrits entre <font color=blue> parenthèses </font>
- Les tuples sont utilisés pour stocker plusieurs éléments 
  dans une seule variable.
- Le tuple est l'un des quatre types de données intégrés à 
  Python utilisés pour stocker des collections de données, 
  les trois autres sont :
    - list, 
    - set,
    - dictionary,
</pre>


- Un tuple est une collection <font color=red><b>ordonnée et immuable</b></font>.
- Les tuples sont plus rapides à créer et à parcourir que les listes, car ils sont plus légers en mémoire (optimisés pour la lecture).

```python
import timeit

print(timeit.timeit("(1, 2, 3, 4, 5)", number=10_000_000))  # tuple
print(timeit.timeit("[1, 2, 3, 4, 5]", number=10_000_000))  # list
```

```python
import dis # dis : pour disassemble

dis.dis("(1, 2, 3, 4, 5)")
dis.dis("[1, 2, 3, 4, 5]") # instruction spéciale BUILD_LIST pour créer une nouvelle liste
```

- Les tuples peuvent être des clés dans un dictionnaire ou être des éléments dans un set (car ils sont immuables) : **cf TDP**

- Les listes ne le peuvent pas (elles sont non-hashables, car elles **ne sont pas immuables**).

```python
mytuple = ("Apple", "orange", "Kiwi")

print(mytuple)
print(type(mytuple))
```

# <font color=blue>Dictionnaires :</font>

- **Non-ordonnées** (Non séquentiels, comme les listes ou chaînes de caractères),
- les valeurs qu'ils contiennent ne sont pas rangées selon un index, mais suivant un identifiant **exclusif** (<u>discriminatoire</u>) : **clé** (unique).
<br>&nbsp;
- Entité de correspondance (mapping), 
- Permet d’__<font color=blue>associer</font>__ une valeur (simple, ou composée) à une clé: <font color=blue>(clé , val)</font> 
    - Appelés  tableaux   __<font color=blue>associatifs</font>__
    - Analogie avec un dictionnaire linguistique : accéder à une définition / mot


- **Analogie** :

```python
# clé entière
L = [5, 4, 22]
print(L[1])
```

### Récupérer une valeur / clé : méthode get()


```python
# print(b)
b = {}
b = {"nom": "BenAhmed", "prenom" : "Nabil", "poids": 70, "taille": 1.73}

name = b.get("nom")
# name = b.get("name")

print(name)
```

```python
import time
```

```python
# Pour de grandes tailles
Lt1 = list()
deb = time.time()
for i in range(1_000_000):
    Lt1.append(i**2)
fin = time.time()

print(f"Liste de {len(Lt1)} elem, tps :",fin-deb)
```

```python
# Première amélioration
Lt2 = list()
deb_ = time.time()
for i in range(1_000_000):
    Lt2.append(i*i)
fin_ = time.time()

print(f"Liste de {len(Lt2)} elem, tps :",fin_ - deb_)
```

```python
# List comprehension
deb_c = time.time()
LCarC = [i*i for i in range(1_000_000)]
fin_c = time.time()
tps = fin_c - deb_c

print(f"Liste de {len(LCarC)} elem, tps :",tps)
```

```python
# List comprehension : temps moyen
tps = 0
nbRepet = 10
for k in range(nbRepet):
    deb_c = time.time()
    LCarC = [i*i for i in range(1_000_000)]
    fin_c = time.time()
    tps = tps + fin_c - deb_c

print(f"Liste de {len(LCarC)} elem, tps :",tps/nbRepet)
```

```python
import time
import random

# Générer une liste de 1 million d'entiers
taille = 10**6
liste = list(range(taille))
dictionnaire = {i: True for i in liste}  # table de hachage

# Choisir un élément à rechercher (par exemple, un élément en fin de structure)
element_a_rechercher = taille - 1

# Recherche dans la liste
debut_liste = time.time()
present_dans_liste = element_a_rechercher in liste
fin_liste = time.time()

# Recherche dans le dictionnaire (table de hachage)
debut_dict = time.time()
present_dans_dico = element_a_rechercher in dictionnaire
fin_dict = time.time()

# Afficher les résultats
print(f"Résultat dans liste       : {present_dans_liste}")
print(f"Temps recherche liste     : {fin_liste - debut_liste:.6f} secondes")

print(f"Résultat dans dictionnaire: {present_dans_dico}")
print(f"Temps recherche dictionnaire : {fin_dict - debut_dict:.6f} secondes")
```
