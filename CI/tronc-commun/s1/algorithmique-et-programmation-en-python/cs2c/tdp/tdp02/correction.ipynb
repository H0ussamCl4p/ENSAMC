Ecrire un programme qui affiche tous les éléments dupliqués d'une liste données sans utiliser aucune fonction prédéfinie en Python.<br>

    * indication : utiliser une double boucle for

```python
# Exemple
# L = [7 , 21 , 5 , 12 , 7 , 91 , 21 ,  12 , 28]
# la sortie est: [7, 21, 12]

L = [7, 21, 5, 12, 7, 91, 21, 12, 28]

d = []   
n = 0
for _ in L:
    n += 1

for i in range(n):
    for j in range(i + 1, n):
        if L[i] == L[j] and L[i] not in d:
            d.append(L[i])
print(d)   
```

Écrire un programme qui prend en paramètre une liste d’entiers T et  et qui affiche la moyenne des éléments de T.


```python
import random
L = [random.randint(0, 10) for i in range(5)]
print(L)
```

```python
# solution 1 : En utilisant une boucle

somme = 0
n = 0
for x in L:       
    somme += x        
    n += 1        
if n == 0:        
    print("None")

moy = somme/n

print(somme)
print("Moyenne =", moy)
```

```python
# solution 2 : En utilisant une sum()
somme=sum(L)
n=len(L)

moy=somme/n

print(somme)
print("Moyenne =", moy)

```

<pre>
Soit T une liste d’entiers. 
Écrire un programme qui met tous les éléments pairs au début de la liste et tous les éléments impairs à la fin de la liste.

<b>Exemple</b>:
L = [5, 6, 9, 8, 10]
L devient :
L = [6, 8, 10, 5, 9] ou L = [8, 6, 10, 9, 5]

<b>N.B.</b>
    - l'ordre interne des paires ou impaires n'est pas interessant
    - Ne pas utiliser d'espaces auxiliaires (l'arrangement doit se faire <b><u>inPlace</u></b>)
</pre>


```python
import random
L = [random.randint(0, 20) for i in range(10)]
L
```

```python
# ecrire votre solution ici

gauche = 0
droite = len(L) - 1

while gauche < droite:
   
    while gauche < droite and L[gauche] % 2 == 0:
        gauche += 1
    
    while gauche < droite and L[droite] % 2 == 1:
        droite -= 1
    
    if gauche < droite:
        L[gauche], L[droite] = L[droite], L[gauche]
    
```

```python
print(L)     
```

- Écrire un programme qui supprime les espaces (s'ils existent) au début d'une chaine de caractère passée en paramètre
        - Ne pas utiliser la fonction strip 

```python
# ecrire votre solution ici

ch = "           test des espaces au debut et fin     "
print(f"ch av :*{ch}*")

# completer la solution ici
# chd = ...

i = 0
while i < len(ch) and ch[i] == " ":
    i += 1

chd = ch[i:]

print(f"ch av :*{chd}*")
```

- Écrire un programme qui supprime les espaces (s'ils existent) à la fin d'une chaine de caractère passée en paramètre
        - Ne pas utiliser la fonction strip

```python
ch = "           test des espaces au debut et fin     "
print(f"ch av :*{ch}*")

i = len(ch) - 1
while i >= 0 and ch[i] == " ":
    i -= 1

chf = ch[:i+1]

print(f"ch ap :*{chf}*")
```
